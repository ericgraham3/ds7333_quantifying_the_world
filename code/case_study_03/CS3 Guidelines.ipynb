## Example of explanation
Files were downloaded to my local Linux Partition and unzipped.  There are 5 folders with ham or spam in the name.
Files in ham are target 1 and files in span are target 0.  I will open each email and get the payload using the email package.  I use the os package to walk the file system. Since there are other files in the directory, I added the if statement to check that only the emails are getting parsed.  I end up with two lists, one with the paths of all the emails and the other is the target variable.

```python
import os
count = 0
targets = []
file_paths = []
for root, dirs, files in os.walk("/home/rdslater/projects/qtw/week5/"): #walking the directory
    for file in files:
        path = os.path.join(root,file) # create the whole file path
        if "ham" in root or "spam" in root: #check to make sure it is in the dataset
            file_paths.append(path)
            count+=1
            if "ham" in root:  #target ham=1, spam = 0
                targets.append(1)
            else:
                targets.append(0)
print(count)
```
